## Reading Data from Bronze Layer

```python
df_cal = spark.read.format('csv')\
            .option("header",True)\
            .option("inferSchema",True)\
            .load('abfss://bronze@awtharundatalake.dfs.core.windows.net/AdventureWorks_Calendar')
```

```python
df_cus = spark.read.format('csv')\
            .option("header",True)\
            .option("inferSchema",True)\
            .load('abfss://bronze@awtharundatalake.dfs.core.windows.net/AdventureWorks_Customers')
```

```python
df_prod = spark.read.format('csv')\
            .option("header",True)\
            .option("inferSchema",True)\
            .load('abfss://bronze@awtharundatalake.dfs.core.windows.net/AdventureWorks_Products')
```

```python
df_ret = spark.read.format('csv')\
            .option("header",True)\
            .option("inferSchema",True)\
            .load('abfss://bronze@awtharundatalake.dfs.core.windows.net/AdventureWorks_Returns')
```

```python
df_prodcat = spark.read.format('csv')\
            .option("header",True)\
            .option("inferSchema",True)\
            .load('abfss://bronze@awtharundatalake.dfs.core.windows.net/AdventureWorks_Product_Categories')
```

```python
df_prodsub = spark.read.format('csv')\
            .option("header",True)\
            .option("inferSchema",True)\
            .load('abfss://bronze@awtharundatalake.dfs.core.windows.net/Product_Subcategories')
```

```python
df_terr = spark.read.format('csv')\
            .option("header",True)\
            .option("inferSchema",True)\
            .load('abfss://bronze@awtharundatalake.dfs.core.windows.net/AdventureWorks_Territories')
```

```python
df_sales = spark.read.format('csv')\
            .option("header",True)\
            .option("inferSchema",True)\
            .load('abfss://bronze@awtharundatalake.dfs.core.windows.net/AdventureWorks_Sales*')
```

## Importing Libraries

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

##Transformations

###Calender

```python
df_cal = df_cal.withColumn('Month',month(col('Date')))\
            .withColumn('Year',year(col('Date')))
df_cal.display()
```

```python
df_cal.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@awtharundatalake.dfs.core.windows.net/AdventureWorks_Calendar")\
            .save()
```

###Customers

```python
df_cus = df_cus.drop('Prefix')\
    .withColumn('FullName', concat_ws(' ', col('FirstName'), col('LastName')))
display(df_cus)
```

```python
df_cus.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@awtharundatalake.dfs.core.windows.net/AdventureWorks_Customers")\
            .save()
```

##Products

```python
df_prod = df_prod.withColumn('ProductSKU', split(col('ProductSKU'), '-')[0])\
                 .withColumn('ProductPrice', round(col('ProductPrice'), 2))\
                 .withColumn('ProductCost', round(col('ProductCost'), 2))\
                 .withColumn('Profit', round(col('ProductPrice') - col('ProductCost'), 2))
display(df_prod)
```

```python
df_prod.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@awtharundatalake.dfs.core.windows.net/AdventureWorks_Products")\
            .save()
```

##Product Categories

```python
df_prodcat.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@awtharundatalake.dfs.core.windows.net/AdventureWorks_Product_Categories")\
            .save()
```

##Product Subcategories

```python
df_prodsub.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@awtharundatalake.dfs.core.windows.net/AdventureWorks_Product_SubCategories")\
            .save()
```

##Returns

```python
df_ret.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@awtharundatalake.dfs.core.windows.net/AdventureWorks_Returns")\
            .save()
```

##Territories

```python
df_terr.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@awtharundatalake.dfs.core.windows.net/AdventureWorks_Territories")\
            .save()
```

##Sales

```python
df_sales.display()
```

```python
from pyspark.sql.functions import to_timestamp, col, year

date_cols = [field.name for field in df_sales.schema.fields if 'date' in field.name.lower()]
for date_col in date_cols:
    df_sales = df_sales.withColumn(date_col, to_timestamp(col(date_col)))

if date_cols:
    df_sales = df_sales.withColumn('SalesYear', year(col(date_cols[0])))

display(df_sales)
```

```python
df_sales.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@awtharundatalake.dfs.core.windows.net/AdventureWorks_Sales")\
            .save()
```
